```python
import pandas as pd

fuel = pd.read_csv("data/fuel.csv")
fuel
```

```python
print(fuel.head())
print(fuel.info())
print(fuel.isnull().sum())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.plot(fuel["FUELCONSUMPTION_COMB"].head(20), label="FUELCONSUMPTION_COMB")
plt.title("laczne zuzycie paliwa dla pierwszych 20 samochodow")
plt.xlabel("samochod")
plt.ylabel("zuzycie")
plt.legend()

plt.savefig("img/lab1_1.png")
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.scatter(fuel["ENGINESIZE"].head(20), fuel["CO2EMISSIONS"].head(20) , label="CO2EMISSIONS")
plt.title("pojemnosc, a co2")
plt.xlabel("silnik")
plt.ylabel("co2")
plt.legend()

plt.savefig("img/lab1_2.png")
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.hist(fuel["CO2EMISSIONS"].head(20), label="CO2EMISSIONS")
plt.title("rozkład emisji CO2")
plt.xlabel("samochod")
plt.ylabel("co2")
plt.legend()

plt.savefig("img/lab1_3.png")
plt.show()
```
